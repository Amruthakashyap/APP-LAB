```python
try:
    a=int(input("enter a number"))
    b=int(input("enter a number"))
    print("Result:", a/b)
except ValueError:
    print("Invalid input. Please enter a valid number.")
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")
```

```python
#multiple exception blocks
try:
    print(x)
except NameError:
    print("Variable x is not defined.")
except:
    print("An exception occurred")
```

```python
#handle multiple exception together
try:
    number=int(input("enter a number"))
    print(10/number)
except (ValueError, ZeroDivisionError) as error:
    print("An error occurred:", error)
```

```python
#else keyword to definne a block of code be ececutedif no exception were raised
try:
    print("Hello")
except:
    print("An exception occurred")
else:
    print("No exception occurred")
```

```python
try:
    number=int(input("enter a number"))
    print(10/number)
except(ValueError):
    print("Invalid input")
except(ZeroDivisionError):
    print("Division by zero is not allowed")
else:
    print("result:", 10/number)
```

```python
try:
    f=open("student.txt","w")
    try:
        f.write("Lourum Ipsum")
    except:
        print("Error: cannot write to file")
    finally:
        f.close()
except:
    print("something went wrong while opening the file")
```
